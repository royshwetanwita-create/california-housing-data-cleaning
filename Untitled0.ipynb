```python
import pandas as pd

df = pd.read_csv("/content/sample_data/california_housing_train.csv")

print(df.head())
```

```python
print("Number of rows and columns:", df.shape)

print("\nColumn names:")
print(df.columns)

print("\nData types:")
print(df.dtypes)
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
print(df.describe().T)
```

```python
# Detecting outliers using the IQR method

Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_count = ((df < lower_bound) | (df > upper_bound)).sum()

print("Number of outliers in each column:")
print(outlier_count)
```

```python
import matplotlib.pyplot as plt

columns = [
    "total_rooms",
    "total_bedrooms",
    "population",
    "households",
    "median_income",
    "median_house_value"
]

for column in columns:
    plt.figure(figsize=(8, 4))
    plt.boxplot(df[column])
    plt.title("Boxplot of " + column)
    plt.ylabel(column)
    plt.show()
```

```python
# Handling outliers using IQR capping

columns = [
    "total_rooms",
    "total_bedrooms",
    "population",
    "households",
    "median_income",
    "median_house_value"
]

for column in columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df[column] = df[column].clip(
        lower=lower_bound,
        upper=upper_bound
    )

print("Outlier treatment completed.")
```

```python
# Checking outliers after treatment

for column in columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = ((df[column] < lower_bound) |
                (df[column] > upper_bound)).sum()

    print(column, ":", outliers)
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
print("Minimum value of each column:")
print(df.min())

print("\nMaximum value of each column:")
print(df.max())
```

```python
# Load the original dataset again
original_df = pd.read_csv("/content/sample_data/california_housing_train.csv")

# Create a copy for the cleaned dataset
cleaned_df = original_df.copy()

# Handle outliers using IQR capping
columns = [
    "total_rooms",
    "total_bedrooms",
    "population",
    "households",
    "median_income",
    "median_house_value"
]

for column in columns:
    Q1 = cleaned_df[column].quantile(0.25)
    Q3 = cleaned_df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    cleaned_df[column] = cleaned_df[column].clip(
        lower=lower_bound,
        upper=upper_bound
    )

print("Original dataset shape:", original_df.shape)
print("Cleaned dataset shape:", cleaned_df.shape)
print("Missing values after cleaning:", cleaned_df.isnull().sum().sum())
print("Duplicate rows after cleaning:", cleaned_df.duplicated().sum())
```

```python
print("BEFORE CLEANING")
print("Rows:", original_df.shape[0])
print("Columns:", original_df.shape[1])
print("Missing values:", original_df.isnull().sum().sum())
print("Duplicate rows:", original_df.duplicated().sum())

print("\nAFTER CLEANING")
print("Rows:", cleaned_df.shape[0])
print("Columns:", cleaned_df.shape[1])
print("Missing values:", cleaned_df.isnull().sum().sum())
print("Duplicate rows:", cleaned_df.duplicated().sum())
```

```python
from sklearn.preprocessing import StandardScaler

# Create a copy for preprocessing
preprocessed_df = cleaned_df.copy()

# Select numerical columns
numerical_columns = preprocessed_df.columns

# Apply StandardScaler
scaler = StandardScaler()

preprocessed_df[numerical_columns] = scaler.fit_transform(
    preprocessed_df[numerical_columns]
)

print("Preprocessing completed!")
print(preprocessed_df.head())
```

```python
# Save the cleaned dataset
cleaned_df.to_csv("cleaned_california_housing.csv", index=False)

# Save the preprocessed dataset
preprocessed_df.to_csv("preprocessed_california_housing.csv", index=False)

print("Both files have been saved successfully!")
```
